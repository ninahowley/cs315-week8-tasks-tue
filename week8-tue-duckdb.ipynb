### Q8: Time frame for a given country

Write a function that given a country name prints out the start and end date of their rankings.

Once you have created the function, call it for a few countries to test it.

```python
def show_timeframe(con, country):
    """A function to print out the time frame a country was in the Netflix rankings using DuckDB."""
    res = con.sql("""
        SELECT MIN(week), MAX(week)
        FROM (
            SELECT country_name, week FROM tvseries
        )
        WHERE country_name = $1
    """, params=[country]).fetchone()
    
    start, end = res[0], res[1]
    print(f"Netflix rankings for {country} exist between {start} and {end}.")

show_timeframe(con, "Chile")
show_timeframe(con, "Russia")
```

### Q10: Country with fewest weeks

Find the country with the fewest weeks in the Netflix rankings.

```python
con.sql("""
SELECT country_name, COUNT(DISTINCT week) AS weekCnt
FROM tvseries
GROUP BY country_name
ORDER BY weekCnt ASC
LIMIT 5;""")
```

## Bonus task: rank2score and kdrama rankings_list

```python
def rank2score(rank: int):
    rank_dict = {
        1:25, 2:18, 3:15, 4:12, 5:10,
        6:8, 7:6, 8:4, 9:2, 10:1
    }
    return rank_dict.get(rank, None)
```

```python
con.create_function(
    "rank2score", rank2score, parameters=["BIGINT"], return_type="BIGINT"
)

con.execute(
    """
    CREATE OR REPLACE TABLE tvseries AS 
    SELECT 
        *, 
        rank2score(weekly_rank) AS score 
    FROM tvseries;
"""
)
```

```python
def get_show_total_score(show, country):
    data = con.sql(f"""SELECT * FROM tvseries WHERE show_title = '{show}' AND country_name = '{country}'""").fetchall()
    total_score = 0
    total_weeks = len(data)
    rankings_list = []
    for d in data:
        total_score += d[7]
        rankings_list.append(d[3])
    show_country_dict = {"total_score":total_score, "total_weeks":total_weeks, "rankings_list":rankings_list}
    return show_country_dict
```

```python
print(get_show_total_score("Squid Game", "United States"))
```

```python
def get_show_total_score2(show):
    countries = con.sql(f"""SELECT DISTINCT country_name FROM tvseries WHERE show_title = '{show}'""").fetchall()
    all_data = []
    for country in countries:
        country = country[0]
        data = con.sql(f"""SELECT * FROM tvseries WHERE show_title = '{show}' AND country_name = '{country}'""").fetchall()
        total_score = 0
        total_weeks = len(data)
        rankings_list = []
        for d in data:
            total_score += d[7]
            rankings_list.append(d[3])
        show_country_dict = {"country_name":country, "total_score":total_score, "total_weeks":total_weeks, "rankings_list":rankings_list}
        all_data.append(show_country_dict)
    df = pd.DataFrame(all_data)
    return df.sort_values(by=["total_score"], ascending=False)
```
